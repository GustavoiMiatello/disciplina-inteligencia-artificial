```python
# @title 20. Matriz Transposta
# Crio uma matriz com algumas linhas e colunas
matriz = [
    [1, 2, 3],
    [4, 5, 6]
]

# Crio uma lista vazia para guardar a matriz transposta
transposta = []

# Percorro as colunas da matriz
for coluna in range(3):
    nova_linha = []

    # Pego os valores de cada linha
    for linha in range(2):
        nova_linha.append(matriz[linha][coluna])

    # Adiciono a nova linha na matriz transposta
    transposta.append(nova_linha)

# Mostro a matriz transposta
print("Matriz transposta:", transposta)
```

```python
# @title 21. Busca Binária
# Crio uma lista já ordenada
numeros = [2, 4, 6, 8, 10, 12, 14, 16, 18]

# Peço o número que o usuário quer procurar
procurado = int(input("Digite o número que deseja procurar: "))

# Defino o início e o fim da lista
inicio = 0
fim = len(numeros) - 1

# Começo considerando que o número não foi encontrado
encontrado = False

# Enquanto ainda houver números para verificar
while inicio <= fim:
    meio = (inicio + fim) // 2

    # Verifico se o número do meio é o procurado
    if numeros[meio] == procurado:
        encontrado = True
        break

    # Se o número procurado for maior, procuro na metade direita
    elif procurado > numeros[meio]:
        inicio = meio + 1

    # Se for menor, procuro na metade esquerda
    else:
        fim = meio - 1

# Mostro o resultado
if encontrado:
    print("Número encontrado!")
else:
    print("Número não encontrado!")
```

Eu fiz dessa forma porque primeiro criei uma lista que já está em ordem crescente. Depois, uso a busca binária para verificar o número que o usuário procura, começando pelo elemento do meio. Dependendo do valor, vou descartando metade da lista e continuo a busca até encontrar o número ou terminar as opções. Escolhi fazer assim porque a busca binária é uma forma mais rápida de procurar um elemento em uma lista ordenada.

```python
# @title 22. Criptografia de Cifra de César
# Crio uma função para criptografar a mensagem
def cifra_cesar(mensagem, deslocamento):
    resultado = ""

    # Passo por cada caractere da mensagem
    for letra in mensagem:
        if letra.isalpha():
            # Transformo a letra em um número e faço o deslocamento
            novo_codigo = ord(letra) + deslocamento

            # Volto para o começo do alfabeto quando necessário
            if letra.islower():
                if novo_codigo > ord('z'):
                    novo_codigo -= 26
                resultado += chr(novo_codigo)
            else:
                if novo_codigo > ord('Z'):
                    novo_codigo -= 26
                resultado += chr(novo_codigo)
        else:
            # Mantenho espaços e outros caracteres iguais
            resultado += letra

    return resultado


# Peço a mensagem e o deslocamento para o usuário
mensagem = input("Digite uma mensagem: ")
deslocamento = int(input("Digite o deslocamento: "))

# Criptografo a mensagem
mensagem_criptografada = cifra_cesar(mensagem, deslocamento)

# Mostro o resultado
print("Mensagem criptografada:", mensagem_criptografada)
```

Eu fiz dessa forma porque primeiro criei uma função para receber a mensagem e o número de posições que cada letra deve andar no alfabeto. Depois, uso o for para passar por cada letra e fazer o deslocamento. Espaços e outros caracteres continuam iguais. No final, mostro a mensagem já criptografada. Escolhi fazer assim porque é uma forma simples de entender como funciona a Cifra de César e também como criar uma função em Python.

```python
# @title 23. Descriptografar Cifra de César
# Crio uma função para descriptografar a mensagem
def cifra_cesar(mensagem, deslocamento):
    resultado = ""

    # Passo por cada caractere da mensagem
    for letra in mensagem:
        if letra.isalpha():
            # Faço o deslocamento para trás
            novo_codigo = ord(letra) - deslocamento

            # Volto para o final do alfabeto quando necessário
            if letra.islower():
                if novo_codigo < ord('a'):
                    novo_codigo += 26
                resultado += chr(novo_codigo)
            else:
                if novo_codigo < ord('A'):
                    novo_codigo += 26
                resultado += chr(novo_codigo)
        else:
            # Mantenho espaços e outros caracteres iguais
            resultado += letra

    return resultado


# Peço a mensagem e o deslocamento para o usuário
mensagem = input("Digite a mensagem criptografada: ")
deslocamento = int(input("Digite o deslocamento usado na criptografia: "))

# Descriptografo a mensagem
mensagem_descriptografada = cifra_cesar(mensagem, deslocamento)

# Mostro o resultado
print("Mensagem descriptografada:", mensagem_descriptografada)
```

Eu fiz dessa forma porque a descriptografia é basicamente o contrário da criptografia. Primeiro, crio uma função para receber a mensagem e o número de posições que as letras devem voltar no alfabeto. Depois, uso o for para passar por cada letra e fazer esse deslocamento. Espaços e outros caracteres continuam iguais. Escolhi fazer assim porque é uma forma de reutilizar a mesma ideia do exercício anterior, mas fazendo o deslocamento para trás.

```python
# @title 24. Validação de CPF
# Crio uma função para validar o CPF
def validar_cpf(cpf):

    # Retiro os pontos e o traço do CPF
    cpf = cpf.replace(".", "").replace("-", "")

    # Verifico se o CPF tem 11 números
    if len(cpf) != 11 or not cpf.isdigit():
        return False

    # Verifico se todos os números são iguais
    if cpf == cpf[0] * 11:
        return False

    # Calculo o primeiro dígito verificador
    soma = 0
    for i in range(9):
        soma += int(cpf[i]) * (10 - i)

    resto = soma % 11

    if resto < 2:
        digito1 = 0
    else:
        digito1 = 11 - resto

    # Calculo o segundo dígito verificador
    soma = 0
    for i in range(10):
        soma += int(cpf[i]) * (11 - i)

    resto = soma % 11

    if resto < 2:
        digito2 = 0
    else:
        digito2 = 11 - resto

    # Verifico se os dois dígitos estão corretos
    if int(cpf[9]) == digito1 and int(cpf[10]) == digito2:
        return True
    else:
        return False


# Peço o CPF para o usuário
cpf = input("Digite o CPF: ")

# Mostro se o CPF é válido ou não
if validar_cpf(cpf):
    print("CPF válido")
else:
    print("CPF inválido")
```

Eu fiz dessa forma porque primeiro criei uma função para receber o CPF e verificar se ele está correto. Primeiro retiro os pontos e o traço e verifico se existem 11 números. Depois, faço os cálculos dos dois dígitos verificadores e comparo com os números do CPF. No final, o programa informa se o CPF é válido ou inválido. Escolhi fazer assim porque é uma forma de separar toda a validação dentro de uma função e deixar o restante do código mais organizado.

```python
# @title 25. Classificação de Números
# Crio uma lista com alguns números
numeros = [10, -5, 0, 8, -3, 0, 15, -10]

# Crio listas para guardar cada tipo de número
positivos = []
negativos = []
zeros = []

# Verifico cada número da lista
for numero in numeros:
    if numero > 0:
        positivos.append(numero)
    elif numero < 0:
        negativos.append(numero)
    else:
        zeros.append(numero)

# Mostro os resultados
print("Números positivos:", positivos)
print("Números negativos:", negativos)
print("Zeros:", zeros)
```

```python
# @title 30. Jogo da Forca
# Palavra que o usuário precisa descobrir
palavra = "python"

# Crio uma lista com os espaços da palavra
descoberta = ["_"] * len(palavra)

# Defino o número de tentativas
tentativas = 6

# Enquanto ainda tiver tentativas
while tentativas > 0 and "_" in descoberta:

    print("\nPalavra:", " ".join(descoberta))
    letra = input("Digite uma letra: ").lower()

    # Verifico se a letra está na palavra
    if letra in palavra:

        # Mostro a letra na posição correta
        for i in range(len(palavra)):
            if palavra[i] == letra:
                descoberta[i] = letra

        print("Você acertou!")

    else:
        tentativas -= 1
        print("Você errou!")

    print("Tentativas restantes:", tentativas)

# Verifico se o usuário ganhou ou perdeu
if "_" not in descoberta:
    print("Parabéns! Você descobriu a palavra:", palavra)
else:
    print("Você perdeu! A palavra era:", palavra)
```

```python
# @title 31. Ordenação por Bolha
# Lista de números que queremos ordenar
numeros = [5, 2, 8, 1, 3]

# Percorre a lista várias vezes
for i in range(len(numeros)):

    # Compara os números que estão lado a lado
    for j in range(len(numeros) - 1):

        # Se o número da esquerda for maior, troca de lugar
        if numeros[j] > numeros[j + 1]:
            numeros[j], numeros[j + 1] = numeros[j + 1], numeros[j]

print("Lista ordenada:", numeros)
```

Eu usei duas estruturas for porque o Bubble Sort precisa comparar os números várias vezes até que todos estejam na ordem correta.

O primeiro for faz com que a lista seja percorrida várias vezes. O segundo for compara os números que estão um ao lado do outro.

A parte numeros[j] > numeros[j + 1] verifica se o número da esquerda é maior que o próximo. Se for, os dois trocam de posição.

Usei numeros[j], numeros[j + 1] = numeros[j + 1], numeros[j] porque essa é uma forma simples do Python de trocar dois valores sem precisar criar outra variável.

Escolhi esse jeito porque o objetivo do exercício é justamente implementar o Bubble Sort, então não usei sort(), que faria a ordenação automaticamente e acabaria fugindo da proposta.

```python
# @title 37. Funções Recursivas
# Função que calcula o fatorial de forma recursiva
def fatorial(numero):

    # Caso base: o fatorial de 0 é 1
    if numero == 0:
        return 1

    # Chama a própria função com um número menor
    return numero * fatorial(numero - 1)


# Pede um número para o usuário
numero = int(input("Digite um número: "))

print("Fatorial:", fatorial(numero))
```

Criei a função fatorial() para calcular o resultado. A diferença é que ela chama a própria função, que é justamente o que caracteriza uma função recursiva.

O if numero == 0 é o caso base. Ele serve para fazer a função parar, já que o fatorial de 0 é 1.

Depois, a função faz numero * fatorial(numero - 1). Por exemplo, se o usuário colocar 5, o cálculo acontece assim:

5 × 4 × 3 × 2 × 1 = 120

Escolhi essa forma porque o exercício pede especificamente uma função recursiva, então não usei um for ou while para calcular o fatorial. A ideia é mostrar como uma função pode chamar ela mesma até chegar ao caso base.

```python
# @title 42. Números Primos até N

n = int(input("Digite um número: "))

for numero in range(2, n + 1):
    primo = True

    for divisor in range(2, numero):
        if numero % divisor == 0:
            primo = False
            break

    if primo:
        print(numero)
```

```python
# @title 48. Ordenação por Inserção

numeros = [5, 2, 4, 1, 3]

for i in range(1, len(numeros)):
    atual = numeros[i]
    j = i - 1

    while j >= 0 and numeros[j] > atual:
        numeros[j + 1] = numeros[j]
        j -= 1

    numeros[j + 1] = atual

print("Lista ordenada:", numeros)
```

No Insertion Sort, considero que a primeira posição já está ordenada.

Depois pego cada próximo número e comparo com os valores anteriores. Enquanto o número anterior for maior, ele é deslocado uma posição para frente.

Por fim, coloco o número atual na posição correta.

```python
# @title 50. Cálculo de Frequência de Letras


texto = input("Digite uma frase: ").lower()

frequencia = {}

for letra in texto:
    if letra.isalpha():
        if letra in frequencia:
            frequencia[letra] += 1
        else:
            frequencia[letra] = 1

print("Frequência das letras:")

for letra, quantidade in frequencia.items():
    print(letra, ":", quantidade)
```

Primeiro usei .lower() para transformar todas as letras em minúsculas. Assim, A e a são consideradas a mesma letra.

Depois percorri cada caractere da frase. O isalpha() verifica se o caractere é uma letra, fazendo com que espaços e outros caracteres não sejam contabilizados.
